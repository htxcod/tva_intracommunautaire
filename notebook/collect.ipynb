```python
import time
import pandas as pd
import requests as req   
import json 

url = "https://ec.europa.eu/taxation_customs/vies/rest-api/ms/{}/vat/{}" 

def appeler(numero):

    reponse = requests.get(url.format(numero[:2], numero[2:]), timeout=30)
    return reponse.json()

```

```python
from pathlib import Path
import sys

module_folder = Path(
    r"C:\Users\Utilisateur\Documents\tva_intracommunautaire\formation-data-IA"
    r"\99-Brief\Data-Analyst\TVA-Intracommunautaire\starter-kit"
)

sys.path.insert(0, str(module_folder))
from validation_structure import normaliser, valider

data_file = module_folder / "data" / "numeros_tva.csv"
df = pd.read_csv(data_file, dtype={"numero_tva": "string"})

```

```python
# sélection au hasard d'un échantillon de 20 numéros valides 
# l'échantillon change à chaque relance

echantillon = df.sample(n=min(20, len(df)))


# Séparer chaque numéro pour l'URL VIES: préfixe du pays et chiffres restants
echantillon["numero_tva_normalise"] = (
    echantillon["numero_tva"].apply(normaliser))

echantillon["pays_code"] = echantillon["numero_tva_normalise"].str[:2]
echantillon["chiffre_tva"] = echantillon["numero_tva_normalise"].str[2:]

print(echantillon[["numero_tva_normalise", "pays_code", "chiffre_tva"]])
```

```python
import requests

# Appel de l'API VIES pour chaque numéro de TVA dans l'échantillon
# Pause d'une seconde entre les appels 

for numero in echantillon["numero_tva_normalise"]:
     resultat = appeler(numero)
     print(f"Numéro: {numero}, Résultat: {resultat}")
     time.sleep(1) 
```
