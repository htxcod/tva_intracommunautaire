```python
import logging
import time
import requests

# 1. forcer l'affichage du résultat sur le notebook

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
    handlers=[
        logging.FileHandler("vies_journal.log", encoding="utf-8"),
        logging.StreamHandler() # Optional: keeps printing in your notebook too
    ],
    force=True 
)

# 2. Log début de campagne

logging.info("=== DÉBUT DE CAMPAGNE ===")

# --- example d'un numero_tva déjà vérifié---

deja_verifie = {"FR27552032534"} # 
echantillon_pairs = [("FR", "27552032534"), ("DE", "129274202")]

for code_pays, numero_tva in echantillon_pairs:
    chiffre_tva = f"{code_pays}{chiffre_tva}"

    # 3. Log passez numero_tva
    
    if chiffre_tva in deja_verifie:
        logging.info(f"PASSEZ | {chiffre_tva} | Déjà vérifié")
        continue

    # 4. mesure du temps de réponse et le status du log HTTP
    
    url = (
                "https://ec.europa.eu/taxation_customs/vies/"
                f"rest-api/ms/{code_pays}/vat/{chiffre_tva}"
            )
    start = time.perf_counter()
    response = requests.get(url, timeout=30)
    duration_ms = round((time.perf_counter() - start) * 1000)

    logging.info(f"APPEL | {numero_tva} | HTTP {response.status_code} | Temps: {duration_ms} ms")

    # 5. les erreurs du service log retournées par VIES
    

    data = response.json()
    if data.get("userError") != "VALID":
        logging.warning(f"ERREUR SERVICE | {numero_tva} | {data.get('userError')}")

# 6. Log campaign end
logging.info("=== FIN DE CAMPAGNE ===")
```
